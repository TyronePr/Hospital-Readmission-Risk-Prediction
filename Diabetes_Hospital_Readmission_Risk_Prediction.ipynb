Why is this important?

1. Hospital readmissions cost the US healthcare system $26 billion annually (CMS).
2. Hospitals face penalties if readmission rates are high for conditions like diabetes, heart failure, and COPD.
3. Early identification of high-risk patients allows for targeted interventions (e.g., follow-up calls, better discharge planning).
